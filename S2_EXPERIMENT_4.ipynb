#DIVERGENCE AND CURL OF A VECTOR FIELD AND ITS VISUALIZATION
**CASE 2**

FIND THE DIVEGENCE AND CURL OF A VECTOR FIELD $\vec(f)=xyi+yj$ and hence visualize the vector field .divergenc and curl

```
import numpy as np
import matplotlib.pyplot as plt
import sympy as sp
from mpl_toolkits.mplot3d import Axes3D

#DEFINE SYMBOLS
x,y,z = sp.symbols('x,y,z')

#DEFINE VECTOR FIELD F=(xy,y,0)
f=sp.Matrix([x*y,y,0])

#COMPUTE DIVERGENCE CORECTLY
div_F = sp.diff(f[0],x)+sp.diff(f[1],y)+sp.diff(f[2],z)
print("div:",div_F)

#COMPUTE CURL

curl_fx=sp.diff(f[2],y)-sp.diff(f[1],z)
curl_fy=sp.diff(f[0],z)-sp.diff(f[2],x)
curl_fz=sp.diff(f[1],x)-sp.diff(f[0],y)

curl=sp.Matrix([curl_fx,curl_fy,curl_fz])
print("curl:",curl)

#CONVERT TO NUMERICAL FUNCTIONS

div_f=sp.lambdify((x,y,z),div_F,"numpy")

curl_fx_f=sp.lambdify((x,y,z),curl_fx,"numpy")
curl_fy_f=sp.lambdify((x,y,z),curl_fy,"numpy")
curl_fz_f=sp.lambdify((x,y,z),curl_fz,"numpy")

#CREATE A 3D GRID

x_vals=np.linspace(-2,2,8)
y_vals=np.linspace(-2,2,8)
z_vals=np.linspace(-2,2,8)

x,y,z=np.meshgrid(x_vals,y_vals,z_vals, indexing='ij') # Added indexing='ij' for consistency with sympy

#COMPUTE VECTOR FIELD COMPONENTS

U,V,W=x*y,y,np.zeros_like(z)

 #COMPUTE DIVERGENCE

Div =div_f(x,y,z)
 #COMPUTE CURL

Curl_x=curl_fx_f(x,y,z)
curl_y=curl_fy_f(x,y,z)
curl_z=curl_fz_f(x,y,z)

 #PLOT 3D VECTOR FIELD
fig=plt.figure(figsize=(15,5))
ax=fig.add_subplot(131,projection='3d')
ax.quiver(x,y,z,U,V,W,color='blue',label='Vector Field')
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')
ax.set_title('Vector Field')
 #DIVERGENCE AS A SURFACE
ax2=fig.add_subplot(132,projection='3d')
# Plotting a 2D slice of the 3D divergence field
ax2.plot_surface(x[:,:,0],y[:,:,0],Div[:,:,0],cmap='coolwarm',edgecolor='b')
ax2.set_xlabel('X')
ax2.set_ylabel('Y')
ax2.set_zlabel('Divergence')
ax2.set_title('Divergence')
 #CURL FIELD
ax3=fig.add_subplot(133,projection='3d')
ax3.quiver(x,y,z,Curl_x,curl_y,curl_z,color='red',label='Curl Field')
ax3.set_xlabel('X')
ax3.set_ylabel('Y')
ax3.set_zlabel('Z')
ax3.set_title('Curl Field')
plt.tight_layout()
plt.show()
```
